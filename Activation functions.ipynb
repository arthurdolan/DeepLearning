```python
def plot_activation(func, name, **kwargs):
    y = func(x, **kwargs) if kwargs else func(x)
    plt.figure(figsize=(6, 4))
    plt.plot(x, y)
    plt.title(name)
    plt.grid(True)
    plt.show()
```

```python
plot_activation(act.relu, "ReLU")
plot_activation(act.heavyside, "Heaviside")
plot_activation(act.sigmoid, "Sigmoid")
plot_activation(act.tanh, "Tanh")
plot_activation(act.prelu, "PReLU", a=0.1)   # Pass parameters when needed
plot_activation(act.leaky_relu, "Leaky ReLU", a=0.01)
plot_activation(act.parameterised_sigmoid, "Parametric Sigmoid", a=0.5)
plot_activation(act.softsign, "Softsign")
```
